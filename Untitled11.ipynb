```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier, export_text
```

```python
df = pd.read_csv("MetroPT3.csv")

df['timestamp'] = pd.to_datetime(df['timestamp'])
df = df.sort_values('timestamp').reset_index(drop=True)
```

```python
features = [
'TP2','TP3','H1','DV_pressure','Reservoirs',
'Oil_temperature','Motor_current','COMP',
'DV_eletric','Towers','MPG',
'Pressure_switch','Oil_level','Caudal_impulses'
]

data = df[features].values
```

```python
train_size = int(len(data) * 0.6)

train_data = data[:train_size]
test_data = data[train_size:]
```

```python
scaler = StandardScaler()

train_scaled = scaler.fit_transform(train_data)
test_scaled = scaler.transform(test_data)
```

```python
def create_windows(data, L=180):
    X = []
    for i in range(L, len(data)):
        X.append(data[i-L:i])
    return np.array(X)

X_train = create_windows(train_scaled, 180)
X_test  = create_windows(test_scaled, 180)
```

```python
import torch
import torch.nn as nn

class LSTMAE(nn.Module):
    def __init__(self, n_features):
        super().__init__()

        self.encoder = nn.LSTM(n_features, 64, batch_first=True)
        self.latent  = nn.LSTM(64, 32, batch_first=True)

        self.decoder = nn.LSTM(32, 64, batch_first=True)
        self.output  = nn.Linear(64, n_features)

    def forward(self, x):
        x, _ = self.encoder(x)
        x, (h, c) = self.latent(x)

        latent = x[:, -1, :].unsqueeze(1)
        latent = latent.repeat(1, x.size(1), 1)

        x, _ = self.decoder(latent)
        x = self.output(x)
        return x
```

```python
model = LSTMAE(n_features=X_train.shape[2])

criterion = nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)

X_train_tensor = torch.tensor(X_train, dtype=torch.float32)

for epoch in range(20):
    model.train()

    optimizer.zero_grad()
    recon = model(X_train_tensor)

    loss = criterion(recon, X_train_tensor)
    loss.backward()
    optimizer.step()

    print(f"Epoch {epoch} | Loss: {loss.item():.6f}")
```

```python
model.eval()

with torch.no_grad():
    train_recon = model(X_train_tensor)
    train_errors = ((X_train_tensor - train_recon)**2).mean(dim=(1,2)).numpy()

    X_test_tensor = torch.tensor(X_test, dtype=torch.float32)
    test_recon = model(X_test_tensor)
    test_errors = ((X_test_tensor - test_recon)**2).mean(dim=(1,2)).numpy()
```

```python
mu = train_errors.mean()
sigma = train_errors.std()

tau = mu + 3 * sigma
print("Threshold:", tau)
```

```python
alpha = 0.1

p = []
z = 0

for e in test_errors:
    y = 1 if e > tau else 0
    z = z + alpha * (y - z)
    p.append(z)

p = np.array(p)
```

```python
plt.figure(figsize=(12,4))
plt.plot(p, label="Failure Probability")
plt.axhline(0.5, color='red')
plt.title("Anomaly Detection Signal")
plt.legend()
plt.show()
```

```python
def aggregate_window(w):
    return [
        w.mean(),
        w.std(),
        w.max(),
        w.min()
    ]

def build_rule_features(X):
    return np.array([
        np.concatenate([aggregate_window(w[:,i]) for i in range(w.shape[1])])
        for w in X
    ])

X_rules_train = build_rule_features(X_train)
X_rules_test  = build_rule_features(X_test)
```

```python
y_rules_train = (train_errors > tau).astype(int)
y_rules_test  = (test_errors > tau).astype(int)
```

```python
tree = DecisionTreeClassifier(max_depth=4, min_samples_leaf=50)

tree.fit(X_rules_train, y_rules_train)

rules = export_text(tree)
print(rules)
```

```python
sensor_names = [
'TP2','TP3','H1','DV_pressure','Reservoirs',
'Oil_temperature','Motor_current','COMP',
'DV_eletric','Towers','MPG',
'Pressure_switch','Oil_level','Caudal_impulses'
]

importances = tree.feature_importances_

grouped = importances.reshape(len(sensor_names), 4)
sensor_importance = grouped.sum(axis=1)

plt.figure(figsize=(10,4))
plt.bar(sensor_names, sensor_importance)
plt.xticks(rotation=45)
plt.title("Sensor Importance")
plt.show()
```
